```python
import pandas as pd
import numpy as np
```

```python
# Load the cleaned dataset

df = pd.read_csv('../data/processed/api_logs_v1_cleaned_2026-05-19.csv')
df.head()
```

```python
df.info()
```

```python
df.columns
```

```python
df.iloc[0]['raw_system_prompt']
```

```python
# Extract 'device_type', 'http_status', and 'user_query' from the raw prompt string

df[['device_type', 'http_status', 'user_query']] = df['raw_system_prompt'].str.split('|', expand=True)
```

```python
df.head()
```

```python
# Remove the original 'raw_system_prompt' column as it is no longer needed

df.drop(columns=['raw_system_prompt'], inplace=True)
```

```python
df.head()
```

```python
# Clean 'device_type': remove prefix and extra spaces

df['device_type'] = df['device_type'].str.replace('Req_SRC:', '').str.strip()
```

```python
# Clean 'http_status': remove prefix, brackets, and extra spaces

df['http_status'] = df['http_status'].str.replace('STATUS:', '').str.replace('[', '').str.replace(']', '').str.strip()
```

```python
# Clean 'user_query': remove prefix and extra spaces

df['user_query'] = df['user_query'].str.replace('Msg: ', '').str.strip()
```

```python
df.head()
```

```python
# Convert 'timestamp' column to datetime format, handling errors

df['timestamp'] = pd.to_datetime(df['timestamp'], errors='coerce')
```

```python
# Extract the 'hour' from the timestamp to analyze time-based patterns

df['hour_of_day'] = df['timestamp'].dt.hour
```

```python
# Drop the original 'timestamp' column to keep the dataframe concise

df.drop(columns=['timestamp'], inplace=True)
```

```python
df.head()
```

```python
df.iloc[0]['backend_server_trace']
```

```python
df.iloc[1]['backend_server_trace']
```

```python
# Clean 'backend_server_trace': remove prefix URL

df['backend_server_trace'] = df['backend_server_trace'].str.replace('Routing -> https://', "").str.strip()
```

```python
df.iloc[1]['backend_server_trace']
```

```python
# Keep only the core server identifier by splitting at the first period

df['backend_server_trace'] = df['backend_server_trace'].apply(lambda x: x.split('.')[0])
```

```python
df.head()
```

```python
# Create a binary target variable 'is_error': 0 for successful requests, 1 for errors

df['is_error'] = np.where(df['http_status'] == 'HTTP_200_OK', 0, 1)
```

```python
df.head()
```

```python
# Calculate and store the character length of each user query

df['query_length'] = df['user_query'].str.len()
```

```python
df.head()
```

```python
# Convert 'device_type' into numerical dummy variables for machine learning models

df = pd.get_dummies(df, columns= ['device_type'], prefix='device', dtype= int)
```

```python
df.head()
```

```python
df.to_csv('../data/processed/api_logs_v1_features_added_2026-05-19.csv', index=False)
```
